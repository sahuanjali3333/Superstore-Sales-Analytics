```python
import pandas as pd

# Dataset load karein
df = pd.read_csv('train.csv')

print("--- 1. DATA SUMMARY ---")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")

# 2. Total Sales nikalne ke liye
total_sales = df['Sales'].sum()
print(f"\nTotal Sales: ₹{total_sales:,.2f}")

# 3. Category-wise Total Sales
print("\n--- 2. CATEGORY-WISE SALES ---")
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
print(category_sales)

# 4. Top 5 Regions by Sales
print("\n--- 3. TOP REGIONS BY SALES ---")
region_sales = df.groupby('Region')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False)
print(region_sales)
```

```python
# Top 10 Cities by Sales
print("--- TOP 10 CITIES BY SALES ---")
top_cities = df.groupby('City')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False).head(10)
print(top_cities)

# Top 10 States by Sales
print("\n--- TOP 10 STATES BY SALES ---")
top_states = df.groupby('State')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False).head(10)
print(top_states)
```

```python
import pandas as pd

# 1. Dataset Load Karein
df = pd.read_csv('train.csv')

print("--- DATA CLEANING PROCESS ---")

# A. Missing Values Check Karein
missing_values = df.isnull().sum()
print("\n1. Missing Values per Column:\n", missing_values[missing_values > 0])

# B. Duplicate Rows Check Karein
duplicates = df.duplicated().sum()
print(f"\n2. Total Duplicate Rows found: {duplicates}")

# C. Agar koi duplicate row ho toh use remove karna:
if duplicates > 0:
    df = df.drop_duplicates()
    print("-> Duplicate rows successfully removed!")

# D. Data Types Check Karein (Jaise Date sahi format mein hai ya nahi)
print("\n3. Current Date Column Data Type:", df['Order Date'].dtype)

# Agar date string mein hai toh use proper Datetime format mein convert karna:
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
print("-> Order Date successfully converted to Datetime format!")
```
